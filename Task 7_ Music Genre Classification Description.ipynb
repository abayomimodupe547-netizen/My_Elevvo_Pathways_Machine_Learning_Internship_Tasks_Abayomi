```python
# Music Genre Classification Notebook
# Two approaches: (A) Tabular MFCC features + classical ML
#                   (B) Mel-spectrogram images + CNN (Keras)
#
# Notes:
# - If you have GTZAN, place it in ./GTZAN (folder structure: GTZAN/genre_name/*.wav)
#   or update `DATA_DIR` below.
# - Required packages: librosa, scikit-learn, numpy, matplotlib, pandas, tensorflow, imblearn (for SMOTE optional)
#   Install example: pip install librosa tensorflow scikit-learn imbalanced-learn
#
# This notebook is written to be runnable even without the real dataset:
# it will generate synthetic numeric features or spectrograms to demonstrate the pipeline.

# -------------------------
# Cell 1: Imports + settings
# -------------------------
import os
import glob
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")

from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix, ConfusionMatrixDisplay
from sklearn.metrics import precision_recall_fscore_support, accuracy_score

# Optional (used later for SMOTE)
try:
    from imblearn.over_sampling import SMOTE
    HAS_SMOTE = True
except Exception:
    HAS_SMOTE = False

# For audio & CNN:
try:
    import librosa
    import librosa.display
    HAS_LIBROSA = True
except Exception:
    HAS_LIBROSA = False

try:
    import tensorflow as tf
    from tensorflow.keras import layers, models, utils, callbacks
    HAS_TF = True
except Exception:
    HAS_TF = False

RANDOM_SEED = 42
np.random.seed(RANDOM_SEED)
random.seed(RANDOM_SEED)

# Path to GTZAN folder (update if your dataset is elsewhere)
DATA_DIR = "./GTZAN"   # expected structure: ./GTZAN/genre_name/*.wav
GENRES = []            # will auto-discover if dataset exists

# -------------------------
# Cell 2: Helper functions
# -------------------------
def list_audio_files(data_dir):
    """Return list of (file_path, genre) pairs for GTZAN-like structure."""
    pairs = []
    if not os.path.isdir(data_dir):
        return pairs
    for genre_dir in os.listdir(data_dir):
        gpath = os.path.join(data_dir, genre_dir)
        if os.path.isdir(gpath):
            for f in glob.glob(os.path.join(gpath, "*.wav")):
                pairs.append((f, genre_dir))
    return pairs

def extract_mfcc_features(y, sr, n_mfcc=20):
    """Extract MFCC and return aggregated summary features (mean + std of each coeff)."""
    mfcc = librosa.feature.mfcc(y=y, sr=sr, n_mfcc=n_mfcc)
    mfcc_mean = np.mean(mfcc, axis=1)
    mfcc_std = np.std(mfcc, axis=1)
    # Concatenate mean and std into one vector
    feat = np.concatenate([mfcc_mean, mfcc_std])
    return feat

def compute_mel_spectrogram_image(y, sr, n_mels=128, fmax=8000, hop_length=512):
    S = librosa.feature.melspectrogram(y=y, sr=sr, n_mels=n_mels, fmax=fmax, hop_length=hop_length)
    S_db = librosa.power_to_db(S, ref=np.max)
    # normalize to 0-1 for image inputs
    S_norm = (S_db - S_db.min()) / (S_db.max() - S_db.min() + 1e-6)
    return S_norm

# -------------------------
# Cell 3: Load dataset (or generate synthetic)
# -------------------------
audio_pairs = list_audio_files(DATA_DIR)
if len(audio_pairs) > 0 and HAS_LIBROSA:
    print(f"Found {len(audio_pairs)} audio files in {DATA_DIR}. Using GTZAN dataset.")
    GENRES = sorted(list({g for _, g in audio_pairs}))
    use_real_audio = True
else:
    print("GTZAN-style dataset not found or librosa unavailable. Falling back to synthetic demo data.")
    use_real_audio = False
    # define synthetic genres
    GENRES = ['classical', 'jazz', 'pop', 'rock', 'metal', 'blues', 'reggae', 'hiphop', 'country', 'disco']

# -------------------------
# Cell 4A: Build MFCC tabular dataset (real audio or synthetic)
# -------------------------
mfcc_features = []
mfcc_labels = []

if use_real_audio:
    # iterate and extract MFCC summary features per file
    for fp, genre in audio_pairs:
        try:
            y, sr = librosa.load(fp, sr=22050, mono=True, duration=30.0)  # load up to 30s
            feat = extract_mfcc_features(y, sr, n_mfcc=20)
            mfcc_features.append(feat)
            mfcc_labels.append(genre)
        except Exception as e:
            print("Failed to load:", fp, e)
else:
    # synthetic numeric features: create 200 samples per genre with MFCC-like dimension (40)
    n_per_genre = 200
    for genre in GENRES:
        for i in range(n_per_genre):
            # create synthetic pattern: base vector for genre + noise
            base = np.sin(np.linspace(0, 3*np.pi, 40)) * (GENRES.index(genre)%5 + 1)
            noise = np.random.normal(0, 0.5, 40)
            feat = base + noise
            mfcc_features.append(feat)
            mfcc_labels.append(genre)
            
X_mfcc = np.array(mfcc_features)
y_mfcc = np.array(mfcc_labels)

print("MFCC tabular dataset shape:", X_mfcc.shape, "Labels:", np.unique(y_mfcc).shape)

# -------------------------
# Cell 4B: Quick MFCC EDA: class distribution
# -------------------------
unique, counts = np.unique(y_mfcc, return_counts=True)
plt.figure(figsize=(8,4))
plt.bar(unique, counts)
plt.title("Class distribution (MFCC tabular dataset)")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# -------------------------
# Cell 5A: Tabular pipeline - encode, scale, train/test split
# -------------------------
le = LabelEncoder()
y_enc = le.fit_transform(y_mfcc)
X_train, X_test, y_train, y_test = train_test_split(X_mfcc, y_enc, test_size=0.2, stratify=y_enc, random_state=RANDOM_SEED)
print("Train/Test sizes (tabular):", X_train.shape, X_test.shape)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Optionally handle imbalance with SMOTE if available
if HAS_SMOTE:
    sm = SMOTE(random_state=RANDOM_SEED)
    X_train_bal, y_train_bal = sm.fit_resample(X_train_scaled, y_train)
    print("Used SMOTE. New train distribution:", np.bincount(y_train_bal))
else:
    X_train_bal, y_train_bal = X_train_scaled, y_train
    print("SMOTE not available — using original training set distribution.")

# -------------------------
# Cell 5B: Train classical classifiers
# -------------------------
rf = RandomForestClassifier(n_estimators=200, random_state=RANDOM_SEED)
lr = LogisticRegression(max_iter=500, multi_class='auto', solver='lbfgs', random_state=RANDOM_SEED)

rf.fit(X_train_bal, y_train_bal)
lr.fit(X_train_bal, y_train_bal)

y_pred_rf = rf.predict(X_test_scaled)
y_pred_lr = lr.predict(X_test_scaled)

print("\nRandom Forest — classification report:")
print(classification_report(y_test, y_pred_rf, target_names=le.classes_))

print("\nLogistic Regression — classification report:")
print(classification_report(y_test, y_pred_lr, target_names=le.classes_))

# Confusion matrices
cm_rf = confusion_matrix(y_test, y_pred_rf)
cm_lr = confusion_matrix(y_test, y_pred_lr)

plt.figure(figsize=(10,4))
plt.subplot(1,2,1)
plt.imshow(cm_rf, interpolation='nearest')
plt.title("Confusion Matrix - Random Forest")
plt.xlabel("Predicted")
plt.ylabel("True")
plt.colorbar()
plt.subplot(1,2,2)
plt.imshow(cm_lr, interpolation='nearest')
plt.title("Confusion Matrix - Logistic Regression")
plt.xlabel("Predicted")
plt.ylabel("True")
plt.colorbar()
plt.tight_layout()
plt.show()

# -------------------------
# Cell 6: Tabular cross-validated metrics (focus on precision/recall/f1)
# -------------------------
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_SEED)
for model, name in [(rf, "Random Forest"), (lr, "Logistic Regression")]:
    # compute precision, recall, f1 with cross_val_score via scoring wrappers is verbose; do per-fold evaluation
    precisions, recalls, f1s = [], [], []
    for train_idx, val_idx in skf.split(X_train_scaled, y_train):
        model_clone = model.__class__(**model.get_params())
        model_clone.fit(X_train_scaled[train_idx], y_train[train_idx])
        preds = model_clone.predict(X_train_scaled[val_idx])
        p, r, f, _ = precision_recall_fscore_support(y_train[val_idx], preds, average='macro', zero_division=0)
        precisions.append(p); recalls.append(r); f1s.append(f)
    print(f"\n{name} cross-validated (train folds) — precision: {np.mean(precisions):.3f}, recall: {np.mean(recalls):.3f}, f1: {np.mean(f1s):.3f}")

# -------------------------
# Part B: Image-based approach with Mel-spectrograms (CNN)
# -------------------------
# We'll produce spectrogram images (n_mels x time) per audio sample and train a small CNN.
# If real audio is not available or librosa missing, synthetic spectrogram-like arrays are generated.

# -------------------------
# Cell 7: Build spectrogram dataset
# -------------------------
specs = []
spec_labels = []

if use_real_audio:
    # For speed, we may process a subset if dataset is huge
    max_files = None  # set an integer to limit
    processed = 0
    for fp, genre in audio_pairs:
        if max_files and processed >= max_files: break
        try:
            y, sr = librosa.load(fp, sr=22050, mono=True, duration=30.0)
            S_img = compute_mel_spectrogram_image(y, sr, n_mels=128, fmax=8000, hop_length=512)
            # optionally resize/pad/truncate to a fixed width (time frames)
            # ensure consistent shape: (128, 128) for example by cropping/padding
            if S_img.shape[1] < 128:
                pad_width = 128 - S_img.shape[1]
                S_img = np.pad(S_img, ((0,0),(0,pad_width)), mode='constant')
            elif S_img.shape[1] > 128:
                S_img = S_img[:, :128]
            specs.append(S_img)
            spec_labels.append(genre)
            processed += 1
        except Exception as e:
            print("Failed:", fp, e)
else:
    # synthetic spectrogram-like arrays: create different base patterns per genre
    n_per_genre = 200
    for gi, genre in enumerate(GENRES):
        for i in range(n_per_genre):
            # base pattern: sinusoids + gaussian blobs differing by genre index
            x = np.linspace(0, 3*np.pi*(gi+1), 128)
            base = np.tile(np.sin(x), (128,1))
            noise = np.random.normal(0, 0.3, base.shape)
            S_img = base * (0.3 + 0.1*(gi%3)) + noise
            # normalize 0-1
            S_img = (S_img - S_img.min()) / (S_img.max() - S_img.min() + 1e-6)
            specs.append(S_img)
            spec_labels.append(genre)

specs = np.array(specs)
spec_labels = np.array(spec_labels)
print("Spectrogram dataset shape:", specs.shape, "Labels:", len(np.unique(spec_labels)))

# -------------------------
# Cell 8: Prepare spectrograms for CNN
# -------------------------
# Convert genre labels to integers
le_spec = LabelEncoder()
y_spec = le_spec.fit_transform(spec_labels)

# Add channel dimension (grayscale)
X_spec = specs[..., np.newaxis].astype('float32')  # shape (N, 128, 128, 1)
print("X_spec shape:", X_spec.shape)

# Train-test split
X_train_s, X_test_s, y_train_s, y_test_s = train_test_split(X_spec, y_spec, test_size=0.2, stratify=y_spec, random_state=RANDOM_SEED)
print("Spectrogram train/test:", X_train_s.shape, X_test_s.shape)

# If tensorflow not available, skip CNN training but still produce dataset info
if not HAS_TF:
    print("TensorFlow not available in this environment. Skipping CNN training. You can still use the tabular MFCC pipeline.")
else:
    # -------------------------
    # Cell 9: Build a simple CNN model
    # -------------------------
    n_classes = len(le_spec.classes_)
    inp_shape = X_train_s.shape[1:]
    model = models.Sequential([
        layers.Input(shape=inp_shape),
        layers.Conv2D(16, kernel_size=(3,3), activation='relu', padding='same'),
        layers.MaxPooling2D((2,2)),
        layers.Conv2D(32, kernel_size=(3,3), activation='relu', padding='same'),
        layers.MaxPooling2D((2,2)),
        layers.Conv2D(64, kernel_size=(3,3), activation='relu', padding='same'),
        layers.MaxPooling2D((2,2)),
        layers.Flatten(),
        layers.Dense(128, activation='relu'),
        layers.Dropout(0.3),
        layers.Dense(n_classes, activation='softmax')
    ])
    model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
    model.summary()

    # callbacks
    cb = [callbacks.EarlyStopping(monitor='val_loss', patience=6, restore_best_weights=True)]

    # fit - use small epochs to keep runtime reasonable
    history = model.fit(X_train_s, y_train_s, validation_split=0.15, epochs=30, batch_size=32, callbacks=cb)

    # -------------------------
    # Cell 10: Evaluate CNN
    # -------------------------
    test_loss, test_acc = model.evaluate(X_test_s, y_test_s, verbose=0)
    print(f"CNN Test accuracy: {test_acc:.3f}")

    y_proba_cnn = model.predict(X_test_s)
    y_pred_cnn = np.argmax(y_proba_cnn, axis=1)
    print("CNN classification report:")
    print(classification_report(y_test_s, y_pred_cnn, target_names=le_spec.classes_))

    # confusion matrix
    cm_cnn = confusion_matrix(y_test_s, y_pred_cnn)
    plt.figure(figsize=(6,5))
    plt.imshow(cm_cnn, interpolation='nearest')
    plt.title("Confusion Matrix - CNN")
    plt.xlabel("Predicted")
    plt.ylabel("True")
    plt.colorbar()
    plt.show()

    # plot training curves
    plt.figure()
    plt.plot(history.history['loss'], label='train_loss')
    plt.plot(history.history['val_loss'], label='val_loss')
    plt.title("CNN Loss")
    plt.xlabel("Epoch")
    plt.ylabel("Loss")
    plt.legend()
    plt.show()

    plt.figure()
    plt.plot(history.history['accuracy'], label='train_acc')
    plt.plot(history.history['val_accuracy'], label='val_acc')
    plt.title("CNN Accuracy")
    plt.xlabel("Epoch")
    plt.ylabel("Accuracy")
    plt.legend()
    plt.show()

# -------------------------
# Final notes printed
# -------------------------
print("\nNotebook run complete.")
print("If you have the GTZAN dataset, set DATA_DIR to the folder containing genre subfolders and rerun.")
print("If you want, I can produce a downloadable .ipynb file for this notebook (tell me 'save notebook').")
```
